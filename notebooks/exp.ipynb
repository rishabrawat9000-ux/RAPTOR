```python

df = df.replace([np.inf, -np.inf], np.nan)

df["Flow Byts/s"] = df["Flow Byts/s"].fillna(
    df["Flow Byts/s"].median()
)
```

```python
print("Duplicates before:", df.duplicated().sum())

df = df.drop_duplicates()

print("Duplicates after:", df.duplicated().sum())
```

```python
df.isnull().sum().sort_values(ascending=False).head(20)
```

```python

df = df.replace([np.inf, -np.inf], np.nan)
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    if df[col].isna().sum() > 0:
        df[col] = df[col].fillna(df[col].median())
```

```python
df.isnull().sum().sort_values(ascending=False).head(20)
```

```python
print("Rows:", len(df))
print("Columns:", len(df.columns))
print("Missing values:", df.isnull().sum().sum())
print("Infinite values:", np.isinf(df.select_dtypes(include=np.number)).sum().sum())
```

```python
df["Label"].value_counts()
```

```python
df["Label"] = df["Label"].astype(str).str.strip()
```

```python
df["Label"].value_counts()
```

```python
df["IsAttack"] = (df["Label"] != "Benign").astype(int)
```

```python
df["IsAttack"].value_counts()
```

```python
stage_mapping = {
    "Benign": "Benign",
    "SSH-Bruteforce": "Initial Access",
    "FTP-BruteForce": "Initial Access"
}

df["AttackStage"] = df["Label"].map(stage_mapping)
```

```python
df["AttackStage"].value_counts()
```

```python
df["Timestamp"] = pd.to_datetime(
    df["Timestamp"],
    errors="coerce"
)
```

```python
print("Invalid timestamps:", df["Timestamp"].isna().sum())
print(df["Timestamp"].head())
```

```python
print("Start:", df["Timestamp"].min())
print("End:", df["Timestamp"].max())
```

```python
df = df.sort_values("Timestamp").reset_index(drop=True)
```

```python
from pathlib import Path

processed_dir = Path(r"C:\code\data\procesed")
processed_dir.mkdir(parents=True, exist_ok=True)

output_file = processed_dir / "02-14-2018_cleaned.csv"

df.to_csv(output_file, index=False)

print(f"Saved: {output_file}")
print(f"Shape: {df.shape}")
```
