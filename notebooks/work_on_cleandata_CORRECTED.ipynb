```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
raw_df = pd.read_csv("/home/rishab/code/RAPTOR/data/raw/02-14-2018.csv")
```

```python
raw_df["Timestamp"] = pd.to_datetime(
    raw_df["Timestamp"],
    format="%d/%m/%Y %H:%M:%S",
    errors="coerce"
)

print("Invalid timestamps:", raw_df["Timestamp"].isna().sum())
```

```python
raw_df = raw_df.sort_values("Timestamp").reset_index(drop=True)
```

```python
print("Start:", raw_df["Timestamp"].min())
print("End:", raw_df["Timestamp"].max())
```

```python
numeric_cols = raw_df.select_dtypes(include=np.number).columns

raw_df[numeric_cols] = raw_df[numeric_cols].replace(
    [np.inf, -np.inf],
    np.nan
)
```

```python
raw_df[numeric_cols] = raw_df[numeric_cols].fillna(
    raw_df[numeric_cols].median()
)
```

```python
raw_df = raw_df.drop_duplicates().reset_index(drop=True)
```

```python
print("Shape:", raw_df.shape)
print("Missing:", raw_df.isnull().sum().sum())
print(
    "Infinite:",
    np.isinf(
        raw_df.select_dtypes(include=np.number)
    ).sum().sum()
)
print(
    "Negative network values:",
    (raw_df[NON_NEGATIVE_FEATURES] < 0).sum().sum()
)
print("NaT timestamps:", raw_df["Timestamp"].isna().sum())
```

```python
DATA_PATH = "/home/rishab/code/RAPTOR/data/processed/02-14-2018_cleaned.csv"

df = pd.read_csv(DATA_PATH)

df["Timestamp"] = pd.to_datetime(
    df["Timestamp"],
    format="%Y-%m-%d %H:%M:%S",
    errors="raise"
)

df = df.sort_values("Timestamp").reset_index(drop=True)

print(df.shape)
print(df["Timestamp"].min())
print(df["Timestamp"].max())
```

```python
WINDOW = "5s"

df["time_window"] = df["Timestamp"].dt.floor(WINDOW)

print("Number of windows:", df["time_window"].nunique())
print(df[["Timestamp", "time_window"]].head(10))
```

```python
state_df = df.groupby("time_window").agg(
    {
        # Traffic volume
        "Tot Fwd Pkts": "sum",
        "Tot Bwd Pkts": "sum",
        "TotLen Fwd Pkts": "sum",
        "TotLen Bwd Pkts": "sum",

        # Traffic rates
        "Flow Byts/s": "mean",
        "Flow Pkts/s": "mean",

        # Flow behaviour
        "Flow Duration": "mean",
        "Fwd Pkt Len Mean": "mean",
        "Bwd Pkt Len Mean": "mean",
        "Fwd Pkt Len Std": "mean",
        "Bwd Pkt Len Std": "mean",

        # Timing
        "Flow IAT Mean": "mean",
        "Flow IAT Std": "mean",
        "Flow IAT Max": "max",
        "Flow IAT Min": "min",

        # TCP
        "SYN Flag Cnt": "sum",
        "ACK Flag Cnt": "sum",
        "RST Flag Cnt": "sum",
        "FIN Flag Cnt": "sum",
        "PSH Flag Cnt": "sum",

        # Packet size
        "Pkt Len Min": "mean",
        "Pkt Len Max": "max",
        "Pkt Len Mean": "mean",
        "Pkt Len Std": "mean",
        "Pkt Size Avg": "mean",

        # Network context
        "Dst Port": "nunique",
        "Protocol": "nunique",
    }
).reset_index()
```

```python
df["IsAttack"] = (
    df["Label"].str.strip().str.lower() != "benign"
).astype(np.int8)

print(df["IsAttack"].value_counts())
```

```python
print(df["Label"].value_counts())
```

```python
attack_state = df.groupby("time_window").agg(
    total_flows=("Label", "size"),
    attack_flows=("IsAttack", "sum"),
    IsAttack=("IsAttack", "max")
).reset_index()

attack_state["attack_ratio"] = (
    attack_state["attack_flows"] /
    attack_state["total_flows"]
)
```

```python
state_df = state_df.merge(
    attack_state,
    on="time_window",
    how="left"
)
```

```python
ATTACK_STAGE_MAP = {
    "Benign": "Benign",
    "SSH-Bruteforce": "Initial Access",
    "FTP-BruteForce": "Initial Access",
}
```

```python
df["AttackStage"] = (
    df["Label"]
    .str.strip()
    .map(ATTACK_STAGE_MAP)
    .fillna("Unknown")
)
```

```python
print(df["AttackStage"].value_counts())
```

```python
def dominant_stage(series):
    counts = series.value_counts()

    if "Initial Access" in counts:
        return "Initial Access"

    return "Benign"
```

```python
stage_state = df.groupby("time_window").agg(
    AttackStage=("AttackStage", dominant_stage)
).reset_index()
```

```python
state_df = state_df.drop(
    columns=["AttackStage"],
    errors="ignore"
)

state_df = state_df.merge(
    stage_state,
    on="time_window",
    how="left"
)
```

```python
# =====================================================
# SAVE TEMPORAL STATE DATASET
# =====================================================

import os

TEMPORAL_PATH = "/home/rishab/code/RAPTOR/data/processed/02-14-2018_temporal_states.csv"

os.makedirs(
    os.path.dirname(TEMPORAL_PATH),
    exist_ok=True
)

# Save the actual temporal network states, NOT the raw dataframe.
state_df.to_csv(
    TEMPORAL_PATH,
    index=False
)

print("Saved temporal states:", TEMPORAL_PATH)
print("Shape:", state_df.shape)
print("Columns:", len(state_df.columns))
print(state_df.columns.tolist())

# Required RAPTOR model features
MODEL_FEATURES = [
    "Tot Fwd Pkts", "Tot Bwd Pkts",
    "TotLen Fwd Pkts", "TotLen Bwd Pkts",
    "Flow Byts/s", "Flow Pkts/s", "Flow Duration",
    "Fwd Pkt Len Mean", "Bwd Pkt Len Mean",
    "Fwd Pkt Len Std", "Bwd Pkt Len Std",
    "Flow IAT Mean", "Flow IAT Std", "Flow IAT Max", "Flow IAT Min",
    "SYN Flag Cnt", "ACK Flag Cnt", "RST Flag Cnt",
    "FIN Flag Cnt", "PSH Flag Cnt",
    "Pkt Len Min", "Pkt Len Max", "Pkt Len Mean",
    "Pkt Len Std", "Pkt Size Avg", "Dst Port", "Protocol"
]

missing = [
    feature for feature in MODEL_FEATURES
    if feature not in state_df.columns
]

if missing:
    raise ValueError(f"Missing RAPTOR model features: {missing}")

print("RAPTOR model features: 27/27 found")
print("Temporal dataset ready for five-step forecasting.")
```
